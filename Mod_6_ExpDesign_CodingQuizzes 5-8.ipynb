#Q1: Use X to predict Y many times via regression with different data sets. Use many samples in each prediction. Which is closest to the average coefficient of X if you do the experiment enough times. 

```python
!pip install statsmodels
```

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm

num_experiments = 500
sample_size = 10000
coefficients = []

for i in range(num_experiments):
    # 1. Generate a new sample of data
    difficulty = np.random.uniform(0, 1, (sample_size,))
    speed = np.maximum(np.random.normal(15, 5, (sample_size,)) - difficulty * 10, 0)
    
    # 2. Run regression: X (difficulty) predicts Y (speed)
    X = sm.add_constant(difficulty)
    model = sm.OLS(speed, X).fit()
    
    # 3. Extract the difficulty coefficient and store it
    coefficients.append(model.params[1])

print(np.mean(coefficients))
```
